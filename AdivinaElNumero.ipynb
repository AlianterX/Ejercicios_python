```python
# Primero imprtamos las siguientes librerias:
import random
import openpyxl
import matplotlib.pyplot as dibujo

def almacena(nombre,intento,resultado):
    excel = openpyxl.load_workbook("C:\\EjerciciosPython\\TareaPython.xlsx")
   
    hoja = excel["Hoja1"]
    hoja.append([nombre,intento,resultado])
   
    excel.save("C:\\EjerciciosPython\\TareaPython.xlsx")
   
   
   

def PartidaSolitaria():
    numAleatorio = random.randint(1,1000)
    ganar = False
    intentos = 1
    print("Elige la dificultad: \n1. Facil \n2. Medio \n3. Dificil")
    dificultad = int(input("dificultad: "))
    if not (dificultad == 1 or dificultad == 2 or dificultad == 3):
        print("Opcion no valida ")
        return
   
    numElegido = int(input("voy a decir este numero: "))
   
    if dificultad == 1:
        numIntentos = 20
    elif dificultad == 2:
        numIntentos = 12
    elif dificultad == 3:
        numIntentos = 5    
    while not ganar and intentos < numIntentos:
        if numAleatorio == numElegido:
            print("Es correcto!")
            ganar = True
        elif numAleatorio < numElegido:
            print("el numero es menor")
            numElegido = int(input())
            intentos = intentos + 1
        else:
            print("el numero es mayor")
            numElegido = int(input())
            intentos = intentos + 1
   
    if ganar:
        nombre = str(input("Introduce tu nombre: "))
        print("Has ganado! " + nombre)
        almacena(nombre,intentos,"ganaste")
       
       
    else :
        nombre = str(input("Introduce tu nombre: "))
        print("Has perdido " + nombre)
        almacena(nombre,intentos,"perdiste")

def Partida2Jugadores():
    ganar = False
    intentos = 1
    print("Elige la dificultad: \n1. Facil \n2. Medio \n3. Dificil")
    dificultad = int(input("dificultad: "))
    if not (dificultad == 1 or dificultad == 2 or dificultad == 3):
        print("Opcion no valida ")
        return

    num1 = int(input("El numero  que hay que adivinar: "))
    num2 = int(input("voy a decir este numero: "))
   
    if dificultad == 1:
        numIntentos = 20
    elif dificultad == 2:
        numIntentos = 12
    elif dificultad == 3:
        numIntentos = 5
    while not ganar and intentos < numIntentos:
        if num1 == num2:
            print("Es correcto!")
            ganar = True
        elif num1 < num2:
            print("el numero es menor")
            num2 = int(input())
            intentos = intentos + 1
        else:
            print("el numero es mayor")
            num2 = int(input())
            intentos = intentos + 1
   
    if ganar:
        nombre = str(input("Introduce tu nombre: "))
        print("Has ganado! " + nombre)
        almacena(nombre,intentos,"ganaste")
       
    else :
        nombre = str(input("Introduce tu nombre: "))
        print("Has perdido " + nombre)
        almacena(nombre,intentos,"perdiste")

def Estadistica():
    excel = openpyxl.load_workbook("C:\\EjerciciosPython\\TareaPython.xlsx")
    hoja = excel["Hoja1"]
   
    print("Estadisticas : ")
    nombres = []
    intentos = []

    topeFilas = hoja.rows
    for fila in topeFilas:
        print("nombre: " + str(fila[0].value) + "   intentos: " + str(fila[1].value) + " resultado: " + str(fila[2].value) )
        nombres.append(fila[0].value)
        intentos.append(fila[1].value)

    x = nombres
    y = intentos
    
    dibujo.bar(x,y)
    dibujo.show()

    dibujo.title("Resultados de la partida")
    dibujo.xlabel("nombre")
    dibujo.ylabel("intentos")
   
def AdivinaElNumero():
    while True: # Aqui pongo un bucle infinito, de este modo aunque se acabe la partida, volvemos al menu principal
        print("Elige una de las siguientes opciones : \n1. Partida modo solitario \n2. Partida 2 Jugadores \n3. Estadistica \n4. Salir")
        seleccion = int(input("elige una: " ))
        if seleccion == 1:
            PartidaSolitaria()
        elif seleccion == 2:
            Partida2Jugadores()
        elif seleccion == 3:
            Estadistica()
        elif seleccion == 4:
            print("apagado")
            break
AdivinaElNumero() 
```
